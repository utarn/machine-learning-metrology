## 1. ทำไมต้อง "pin" เวอร์ชัน — บทเรียนจากงานเมโทรโลยีเอง

ในงานสอบเทียบ เราไม่เขียนผลลัพธ์ว่า "แม่นยำพอประมาณ" แต่เขียนว่าค่า reference คืออะไร ความไม่แน่นอนเท่าไร และใช้เครื่องมือรุ่นไหน — เพราะ **ผลลัพธ์ที่ทำซ้ำไม่ได้ คือผลลัพธ์ที่ใช้ไม่ได้**

ซอฟต์เวอร์ก็เหมือนกัน: ถ้าวันจันทร์คุณเทรนโมเดลด้วย scikit-learn 1.9.1 แล้ววันศุกร์เพื่อนรันโค้ดเดียวกันด้วย 1.7 ผลลัพธ์อาจต่างกันเล็กน้อยแต่พอจะทำให้ตัวเลขสอบเทียบใช้ไม่ได้ ทางแก้คือ **environment แบบ pinned** — ล็อกเวอร์ชันทุกตัวไว้ในไฟล์ล็อก แล้วทุกเครื่องติดตั้งชุดเดียวกัน

เครื่องมือในหลักสูตรนี้:

````python
# เวอร์ชันที่หลักสูตรล็อกไว้ (ดูเหตุผลฉบับเต็มที่ docs/research/version-stack.md)
pinned = {
    "python": "3.14.x",
    "numpy": "2.5.3",
    "pandas": "3.0.6",
    "polars": "1.44.2",
    "scikit-learn": "1.9.1",
    "xgboost": "3.4.1",
    "shap": "0.52.0",
    "streamlit": "1.64.0",
}
for name, version in pinned.items():
    print(f"{name:15s} {version}")
````

## 2. ติดตั้ง environment ด้วยสคริปต์ one-click

**Windows (เครื่องในห้องแล็บส่วนใหญ่):** เปิด PowerShell ที่โฟลเดอร์โปรเจกต์ แล้วรัน

```powershell
powershell -ExecutionPolicy ByPass -File .\setup.ps1
```

สคริปต์จะทำ 5 ขั้นตอนให้อัตโนมัติ:
1. ติดตั้ง `uv` (เครื่องมือจัดการ Python และแพ็กเกจ — เร็วกว่า pip มาก)
2. ติดตั้ง CPython 3.14 ผ่าน `uv`
3. `uv sync --frozen` — ติดตั้งไลบรารีตามไฟล์ `uv.lock` **เป๊ะทุกเวอร์ชัน**
4. ลงทะเบียน Jupyter kernel ชื่อ `ml-metrology`
5. รัน `smoke_test.py` — ถ้าขึ้น **ผ่าน (PASS)** แปลว่าเครื่องพร้อม

**macOS / Linux:** ติดตั้ง `uv` แล้วรันสองคำสั่งเดียวกัน:

```bash
curl -LsSf https://astral.sh/uv/install.sh | sh
uv sync --frozen && uv run python smoke_test.py
```

ถ้าห้องแล็บไม่มีอินเทอร์เน็ต ให้ทำตามวิธี offline cache ที่คอมเมนต์หัวไฟล์ `setup.ps1`

## 3. ตรวจว่า environment ของคุณตรงกับที่ล็อกไว้

เซลล์ถัดไปเทียบเวอร์ชันที่ติดตั้งจริงกับค่า pinned — รันแล้วดูว่าทุกบรรทัดเขียว (ตรงกัน) หรือไม่

````python
import sys

import matplotlib
import numpy
import pandas
import polars
import sklearn
import xgboost

installed = {
    "python": sys.version.split()[0],
    "numpy": numpy.__version__,
    "pandas": pandas.__version__,
    "polars": polars.__version__,
    "scikit-learn": sklearn.__version__,
    "xgboost": xgboost.__version__,
    "matplotlib": matplotlib.__version__,
}

expected = {
    "python": "3.14.",
    "numpy": "2.5.3",
    "pandas": "3.0.6",
    "polars": "1.44.2",
    "scikit-learn": "1.9.1",
    "xgboost": "3.4.1",
    "matplotlib": "3.11.2",
}

ok = True
for name, want in expected.items():
    got = installed[name]
    match = got == want if "." not in want.lstrip("3.14") else got.startswith(want)
    print(f"{'OK ' if match else 'MISMATCH'} {name:15s} got {got}  (want {want})")
    ok &= match
print("\nผ่าน (PASS)" if ok else "\nไม่ผ่าน — รัน setup.ps1 อีกครั้ง หรือถามผู้สอน")
````

## 4. เปิด JupyterLab และเลือก kernel ให้ถูก

```bash
uv run jupyter lab
```

- `uv run` หมายถึง "รันด้วย environment ที่ล็อกไว้" — **อย่า**ใช้ `jupyter lab` ตรง ๆ เพราะอาจชน Python ของระบบที่ไลบรารีต่างกัน
- เปิด notebook นี้แล้วเลือก kernel **"Python (ml-metrology)"** มุมขวาบน
- ทุก notebook ในหลักสูตรอ่านข้อมูลจากโฟลเดอร์ `datasets/` ด้วย **relative path** ดังนั้นต้องเปิด JupyterLab **จาก repo root** เท่านั้น

````python
from pathlib import Path

cwd = Path.cwd()
if (cwd / "datasets").exists():
    print("OK — cwd คือ repo root และมองเห็น datasets/ รัน notebook ทุกไฟล์ในหลักสูตรได้")
else:
    print("คำเตือน: cwd นี้ไม่ใช่ repo root:", cwd)
    print("ให้หยุด (Ctrl-C) แล้วรัน 'uv run jupyter lab' จาก repo root")
    print("(notebook อื่นของหลักสูตรมีตัวช่วยค้นหา datasets/ ให้เอง แต่การเปิดจาก root ยังคงเป็นแนวปฏิบัติที่ถูก)")
````
